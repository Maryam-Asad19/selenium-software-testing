## Testfall 1: Bildersuche
(Diese Anwendung sollte mit Mac getestet werde.)

```python
# Öffne Unsplash-Seite
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# WebDriver initialisieren
driver = webdriver.Chrome()
driver.get("https://unsplash.com/de/")
```

```python
# Suche nach "Hochschule"
search_box = driver.find_element(by=By.NAME, value="searchKeyword")
search_box.send_keys("Hochschule")
search_box.send_keys(Keys.RETURN)
time.sleep(5)  # Warten auf die Suchergebnisse
```

```python
# Scroll to load photos
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(5)  # Wait for photos to load

# Locate photo elements
photos = driver.find_elements(by=By.CSS_SELECTOR, value="figure")  # Update selector if needed

# Check if more than 10 photos are displayed
if len(photos) > 10:
    print(f"More than 10 photos found: {len(photos)}")
else:
    print(f"Less than or equal to 10 photos found: {len(photos)}")
```

```python
# Click on the "Filter" button
filter_button = driver.find_element(By.XPATH, "//button[span[text()='Filter']]")
filter_button.click()
print("Filter button clicked.")
time.sleep(2)
```

```python
# Click on the "Neueste" button
neueste_button = driver.find_element(By.XPATH, "//button[div[text()='Neuestes']]")
neueste_button.click()
print("Neueste button clicked.")
```

```python
# Click on the "Anwenden" button
close_button = driver.find_element(By.XPATH, "//button[text()='Anwenden']")
close_button.click()
print("Anwenden button clicked.")
```

```python
# Auswahl des ersten Fotos
try:
    # Sicherstellen, dass das Foto sichtbar ist
    first_photo = WebDriverWait(driver, 10).until(
        EC.visibility_of_element_located((By.CSS_SELECTOR, "figure:nth-of-type(1)"))
    )
    driver.execute_script("arguments[0].scrollIntoView();", first_photo)
    first_photo.click()
    print("Das erste Foto wurde erfolgreich ausgewählt und angeklickt.")
    time.sleep(5)
except Exception as e:
    print(f"Fehler beim Auswählen des ersten Fotos: {e}")
```

```python
# Veröffentlichungsdatum ausgeben
try:
    # Suchen des Veröffentlichungsdatums
    date_element = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.XPATH, "//time"))
    )
    publication_date = date_element.text
    print(f"Veröffentlichungsdatum des ersten Bildes: {publication_date}")
except Exception as e:
    print(f"Fehler beim Abrufen des Veröffentlichungsdatums: {e}")
```

## Testfall 2: Entdecken und herunterladen

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import os
```

```python
# WebDriver initialisieren
driver = webdriver.Chrome()
print("WebDriver wurde erfolgreich initialisiert.")
```

```python
# Unsplash-Seite öffnen
driver.get("https://unsplash.com/de/")
print("Unsplash-Seite wurde geöffnet.")
time.sleep(2)

# 'Entdecken'-Button finden und anklicken
explore_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.LINK_TEXT, "Entdecken"))
)

if explore_button:
    explore_button.click()
    print("Erfolgreich zu 'Entdecken' navigiert.")
    time.sleep(5)
else:
    print("'Entdecken'-Button nicht gefunden. Das Programm wird beendet.")
```

```python
# Alle Fotos auf der Seite finden
photos = WebDriverWait(driver, 10).until(
    EC.presence_of_all_elements_located((By.CSS_SELECTOR, "figure"))
)

if len(photos) == 0:
    print("Keine Fotos gefunden. Das Programm wird beendet.")
else:
    print(f"{len(photos)} Fotos wurden gefunden. Suche nach dem 'Kostenlos herunterladen'-Button...")
    for index, photo in enumerate(photos):
        print(f"Überprüfe Foto {index + 1}...")
        
        # Zum Foto scrollen und anklicken
        driver.execute_script("arguments[0].scrollIntoView();", photo)
        photo.click()
        time.sleep(3)  # Warten, bis das Foto geladen ist
        
        # Nach "Kostenlos herunterladen"-Button suchen
        download_buttons = driver.find_elements(
            By.XPATH, "//a[contains(@data-testid, 'non-sponsored-photo-download-button') and contains(text(), 'Kostenlos herunterladen')]"
        )
        
        if len(download_buttons) > 0:
            # Wenn der Button existiert, anklicken und WebDriver beenden
            download_buttons[0].click()
            print(f"Der 'Kostenlos herunterladen'-Button wurde auf Foto {index + 1} gefunden und angeklickt.")
            time.sleep(5)  # Warten, bis der Download beginnt
            
            driver.quit()
            print("WebDriver wurde nach erfolgreichem Download beendet.")
            break
        else:
            print(f"Kein 'Kostenlos herunterladen'-Button auf Foto {index + 1} gefunden. Zurück zur Galerie...")
            driver.back()
            time.sleep(2)  # Warten, bis die Galerie geladen ist
```

```python
# Überprüfen, ob Download erfolgreich war
download_path = os.path.expanduser("~/Downloads")  # Passen Sie den Pfad an Ihr System an
files = os.listdir(download_path)

if any("unsplash" in file for file in files):
    print("Der Download war erfolgreich.")
else:
    print("Der Download war nicht erfolgreich. Überprüfen Sie den Downloads-Ordner.")
```

## Reflexion über GUI-Testing

### Vorteile

1. **Realitätsnähe**:
   - GUI-Tests simulieren reale Benutzerinteraktionen und decken Probleme auf.

2. **Automatisierung wiederkehrender Aufgaben**:
   - Wiederholbare GUI Tests sparen Zeit und Aufwand.

3. **Visuelle Prüfung**:
   - GUI-Tests können sicherstellen, dass visuelle Elemente korrekt angezeigt und angeordnet werden.

### Nachteile

1. **Instabilität**:
   - GUI-Tests sind anfällig für Änderungen in der Benutzeroberfläche (z.B. Änderung von CSS-Selektoren oder XPath).

2. **Wartungsaufwand**:
   - Anpassungen an Skripten sind erforderlich, wenn sich die Struktur oder das Design der Anwendung ändert.

3. **Abhängigkeit von Umgebungen**:
   - Tests können je nach Betriebssystem, Bildschirmauflösung oder Browserversion unterschiedlich ausfallen.
